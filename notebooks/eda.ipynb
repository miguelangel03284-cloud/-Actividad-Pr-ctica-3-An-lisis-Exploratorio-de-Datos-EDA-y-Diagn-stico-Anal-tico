```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

PATH_FILE = "../data/raw/Titanic.csv"

df = pd.read_csv(PATH_FILE)
df.head()
```

## Resumen de variables del dataset Titanic

| Variable | Definición | Key |
|---|---|---|
| `survival` | Supervivencia | 0 = No, 1 = Sí |
| `pclass` | Clase del pasajero, es decir, categoria de clase | 1 = 1ra, 2 = 2da, 3 = 3ra |
| `sex` | Sexo |  |
| `Age` | Edad en años |  |
| `sibsp` | Número de hermanos / cónyuges a bordo |  |
| `parch` | Número de padres / hijos a bordo |  |
| `ticket` | Número del billete |  |
| `fare` | Tarifa pagada por el pasajero |  |
| `cabin` | Número de cabina |  |
| `embarked` | Puerto de embarque | C = Cherbourg, Q = Queenstown, S = Southampton |

### Descripción breve
- `survival`: indica si el pasajero sobrevivió.
- `pclass`: clase del pasajero.
- `sex`: género del pasajero.
- `Age`: edad en años.
- `sibsp`: cantidad de hermanos o cónyuges a bordo.
- `parch`: cantidad de padres o hijos a bordo.
- `ticket`: identificador del boleto.
- `fare`: costo del pasaje.
- `cabin`: cabina asignada.
- `embarked`: puerto donde abordó.

```python
#Matriz de correlación (Pearson / Spearman) y detección de variables relevantes o redundantes.

variables = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Correlación de Pearson
corr_pearson = df[variables].corr(method='pearson')

sns.heatmap(corr_pearson, annot=True, cmap='coolwarm',
            fmt='.2f', ax=axes[0])

axes[0].set_title('Correlación de Pearson')

# Correlación de Spearman
corr_spearman = df[variables].corr(method='spearman')

sns.heatmap(corr_spearman, annot=True, cmap='coolwarm',
            fmt='.2f', ax=axes[1])

axes[1].set_title('Correlación de Spearman')

plt.tight_layout()
plt.show()
```

Conclusión 1: Clase del pasajero
La variable Pclass presenta una correlación negativa de -0.34 con la supervivencia en ambos métodos. Esto indica que los pasajeros de clases inferiores tuvieron una menor tendencia a sobrevivir que los pasajeros de primera clase.

Conclusión 2: Tarifa pagada
La variable Fare presenta una correlación positiva con la supervivencia, con valores de 0.26 en Pearson y 0.32 en Spearman. Esto sugiere que los pasajeros que pagaron tarifas más elevadas tuvieron una mayor tendencia a sobrevivir.

Conclusión 3: Edad y familiares
Las variables Age, SibSp y Parch presentan correlaciones muy débiles con la supervivencia. Por tanto, individualmente no muestran una relación importante con esta variable según los resultados obtenidos.

```python
fare = df['Fare']
q1 = fare.quantile(0.25)
q3 = fare.quantile(0.75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

print(f"Límite Inferior de Tukey: {limite_inferior:.4f}")
print(f"Límite Superior de Tukey: {limite_superior:.4f}")

outliers_df = df[(fare < limite_inferior) | (fare > limite_superior)]
total_outliers = len(outliers_df)
porcentaje_outliers = (total_outliers / len(df)) * 100

print(f"\nCantidad total de outliers detectados: {total_outliers}")
print(f"Porcentaje de outliers en la muestra: {porcentaje_outliers:.2f}%")

print("\nRegistros identificados como atípicos:")
display(outliers_df)


plt.figure(figsize=(10, 4))
plt.boxplot(df['Fare'], vert=False, patch_artist=True, boxprops=dict(facecolor='lightblue', color='blue'))
plt.title('Diagrama de Caja y Bigotes para Fare')
plt.xlabel('Tarifa ($)')
plt.grid(axis='x', alpha=0.5)
plt.show()

# Se observa que la mayoría de los pasajeros pagaron tarifas bajas (menos de 100), con algunos pasajeros que pagaron tarifas muy altas (más de 200). Esto podría indicar que la mayoría viajaba en tercera clase, mientras que unos pocos viajaban en primera class, hay valores atípicos que podrían ser pasajeros de primera class o de lujo, y los valores se observan encima de los 70 dólares.
```
